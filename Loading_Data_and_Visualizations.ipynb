```python


#tweak_data helps clean data and see steps used to clean data

import yfinance as yf

def fix_cols(df):
    columns = df.columns
    outer = [col[0] for col in columns]
    df.columns = outer
    return df

def tweak_data():
    raw = yf.download('SPY AAPL', start='2010-01-01', end= '2019-12-31')

    return (raw
            .iloc[:, ::2]
            .pipe(fix_cols)
    )

tweak_data()

```
